```python
from sklearn.datasets import load_iris
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN, KMeans
```

```python
iris = load_iris()

x = iris.data
y  =iris.target
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
# visualize
sns.scatterplot(x=x_scaled[:, 0], y=x_scaled[:,2])
```

```python
dbscan = DBSCAN(
    eps = 0.8,
    min_samples = 5
)
labels = dbscan.fit_predict(x)
```

```python
sns.scatterplot(x=x_scaled[:, 0], y=x_scaled[:,2], c = labels)
```

# non_linear data

```python
from sklearn.datasets import make_moons
X,y = make_moons(
    n_samples = 300,
    noise = 0.05,
    random_state = 42
)
```

```python
X_scaled = scaler.fit_transform(X)
```

```python
sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:, 1])
```

```python
# kmeans
kmeans = KMeans(
    n_clusters = 2,
    random_state = 42
)
labels = kmeans.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:, 1], c= labels)
```

```python
# DBSCAN
dbscan = DBSCAN(
    eps = 0.8,
    min_samples = 5
)
labels2 = dbscan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:, 1], c= labels2)
```
