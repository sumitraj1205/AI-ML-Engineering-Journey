```python
from sklearn.ensemble import StackingClassifier, VotingClassifier, RandomForestClassifier, GradientBoostingClassifier
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, recall_score, precision_score,confusion_matrix
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("novagen_dataset.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated()
```

```python
x = df.drop(columns = ["Target"])
y = df["Target"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(
    x,
    y,
    test_size = 0.2,
    random_state = 40
)
```

```python
ss = StandardScaler()
```

```python
x_train_scaled = ss.fit_transform(x_train)
x_test_scaled = ss.transform(x_test)
```

LOgistic Regression

```python
lr = LogisticRegression(
    penalty = "l2",
    solver = "liblinear",
    max_iter = 1000
)
lr.fit(x_train_scaled,y_train)
y_pred_lr = lr.predict(x_test_scaled)
print("Logistic Regression Accuracy:", accuracy_score(y_test, y_pred_lr))
print("Logistic Regression Recall:", recall_score(y_test, y_pred_lr))
print("Logistic Regression precision:", precision_score(y_test, y_pred_lr))
print("Logistic Regression confusion matrix:", confusion_matrix(y_test, y_pred_lr))
print(classification_report(y_test, y_pred_lr))
```

# knn

```python
knn = KNeighborsClassifier(
     n_neighbors=5,
    metric="euclidean"
    
)
knn.fit(x_train_scaled,y_train)
y_pred_knn = knn.predict(x_test_scaled)
print("KNN Accuracy:", accuracy_score(y_test, y_pred_knn))
print("KNN Recall:", recall_score(y_test, y_pred_knn))
print("KNN precision:", precision_score(y_test, y_pred_knn))
print("KNN confusion matrix:", confusion_matrix(y_test, y_pred_knn))
print(classification_report(y_test, y_pred_knn))
```

# Naive bayes

```python
NB = GaussianNB()
NB.fit(x_train_scaled,y_train)
y_pred_NB = NB.predict(x_test_scaled)
print("NB Accuracy:", accuracy_score(y_test, y_pred_NB))
print("NB Recall:", recall_score(y_test, y_pred_NB))
print("NB precision:", precision_score(y_test, y_pred_NB))
print("NB confusion matrix:", confusion_matrix(y_test, y_pred_NB))
print(classification_report(y_test, y_pred_NB))
```

Decision tree

```python
dt = DecisionTreeClassifier(
     max_depth = 4,
    min_samples_leaf = 4,
    min_samples_split = 4,
    random_state = 42
)
dt.fit(x_train_scaled,y_train)
y_pred_dt = dt.predict(x_test_scaled)
print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))
print("Decision Tree Recall:", recall_score(y_test, y_pred_dt))
print("Decision Tree:", precision_score(y_test, y_pred_dt))
print("Decision Tree matrix:", confusion_matrix(y_test, y_pred_dt))
print(classification_report(y_test, y_pred_dt))
```

# Random forest

```python
rf = RandomForestClassifier(
    n_estimators = 201,
    max_depth = 4,
    min_samples_leaf = 4,
    min_samples_split = 4,
    ccp_alpha = 0.01,
    random_state = 42
)
rf.fit(x_train_scaled,y_train)
y_pred_rf = rf.predict(x_test_scaled)
print("Random forest Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Random forest Recall:", recall_score(y_test, y_pred_rf))
print("Random forest:", precision_score(y_test, y_pred_rf))
print("Random forest matrix:", confusion_matrix(y_test, y_pred_rf))
print(classification_report(y_test, y_pred_rf))
```

# Gradient boosting

```python
gb = RandomForestClassifier(
    n_estimators = 201,
    max_depth = 4,
    min_samples_leaf = 4,
    min_samples_split = 4,
    ccp_alpha = 0.01,
    random_state = 42
)
gb.fit(x_train_scaled,y_train)
y_pred_gb = gb.predict(x_test_scaled)
print("Gradient boosting Accuracy:", accuracy_score(y_test, y_pred_gb))
print("Gradient boosting Recall:", recall_score(y_test, y_pred_gb))
print("Gradient boosting:", precision_score(y_test, y_pred_gb))
print("Gradient boosting matrix:", confusion_matrix(y_test, y_pred_gb))
print(classification_report(y_test, y_pred_gb))
```

# Voting Classifier

```python
vc = VotingClassifier(
    estimators = [
        ("lr", lr),
        ("dt", dt),
        ("knn", knn),
        ("NB", NB)
    ],
    voting = "soft"
)
vc.fit(x_train_scaled,y_train)
y_pred_vc = vc.predict(x_test_scaled)
print("Voting Classifier Accuracy:", accuracy_score(y_test, y_pred_vc))
print("Voting Classifier Recall:", recall_score(y_test, y_pred_vc))
print("Voting Classifier:", precision_score(y_test, y_pred_vc))
print("Voting Classifier matrix:", confusion_matrix(y_test, y_pred_vc))
print(classification_report(y_test, y_pred_vc))
```

# Stacking classifier

```python
sc = StackingClassifier(
    estimators = [
        ("lr", lr),
        ("dt", dt),
        ("knn", knn),
        ("NB", NB)
    ],
    final_estimator = dt,
    cv = 5
)
sc.fit(x_train_scaled,y_train)
y_pred_sc = sc.predict(x_test_scaled)
print("Stacking Classifier Accuracy:", accuracy_score(y_test, y_pred_sc))
print("Stacking Classifier Recall:", recall_score(y_test, y_pred_sc))
print("Stacking Classifier:", precision_score(y_test, y_pred_sc))
print("Stacking Classifier matrix:", confusion_matrix(y_test, y_pred_sc))
print(classification_report(y_test, y_pred_sc))
```
