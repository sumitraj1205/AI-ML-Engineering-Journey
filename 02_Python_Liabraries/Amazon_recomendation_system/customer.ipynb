```python
#for importing good dataset
import json
def dataset(filename):
    with open(filename,"r") as f:
        data = json.load(f)
    return data
data = dataset("data.json")
print(data,type(data))
```

```python
#for making a good dataset
def clean_data(data2):
    rating_dic = {"one" : 1,"two" : 2,"three" : 3, "four" : 4, "five" : 5}
    cleaned_data = []
    unique = set()
    for user in data2:
        
        #cleaning of data
        raw_rating = str(user["rating"]).strip().lower()
        
        if (raw_rating in rating_dic):
            raw_rating = rating_dic[raw_rating]
        user["rating"] = raw_rating
        
        #filling null value
        raw_age = user.get("age")
        if raw_age == None:
            user["age"] = None
            
        #deduplication
        if user["name"].strip() in unique:
            continue
        unique.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
good_data = clean_data(data)
```

```python
# for keeping good insight on data
def process(data):
    su = 0
    count = 0
    for users in data:
        # calculating the average
        su += float(users["rating"])
        # percentage of low ratings
        if(float(users["rating"])<3):
            count +=1
    
    print("average rating:",su/len(data))
    print("percentage of low  rating:",count/len(data)*100)
process(good_data)
```

```python
#recomendation
def recomend(data):
    recomendation = []
    for users in data:
        curr_rec = {}
        curr_rec["name"] = users["name"]
        if (float(users["rating"])<=4):
            curr_rec["brand"] = "apple"
        else:
            curr_rec["brand"] = "samsung"
        recomendation.append(curr_rec)
    return recomendation
recomend(good_data)
            
```
