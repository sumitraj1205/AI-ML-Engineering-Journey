```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme()
a = sns.load_dataset("tips")
print(a.head())
sns.lineplot(
    data = a,
    x = "total_bill",
    y = "tip"
)
```

```python
sns.set_theme()
a = sns.load_dataset("tips")
print(a.head())
sns.scatterplot(
    data = a,
    x = "total_bill",
    y = "tip"
)
```
