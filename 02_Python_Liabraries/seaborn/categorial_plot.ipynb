```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme()
a = sns.load_dataset("tips")
print(a.head())
sns.barplot(
    data = a,
    x = "sex",
    y = "tip"
)
```

```python
sns.set_theme()
a = sns.load_dataset("tips")
print(a.head())
sns.boxplot(
    data = a,
    x = "sex",
    y = "tip"
)
```
