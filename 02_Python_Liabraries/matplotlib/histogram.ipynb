```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(0)
score = np.random.normal(70,10,100)
bin = [30,40,50,60,70,80,90,100]
plt.hist(score,color = "skyblue",edgecolor = "black")
plt.xlabel("no. of student")
plt.ylabel("marks")
```

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(0)
score = np.random.normal(70,10,100)
bin = [30,40,50,60,70,80,90,100]
plt.hist(score,color = "skyblue",edgecolor = "black",bins = bin)
plt.xlabel("no. of student")
plt.ylabel("marks")
```
