```python
import matplotlib.pyplot as plt
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]
plt.pie(amounts,labels = expenses)
plt.title("expenses vs amounts")
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]
plt.pie(amounts,labels = expenses,autopct = "%1.1f%%")
plt.title("expenses vs amounts")
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]
plt.pie(amounts,labels = expenses,autopct = "%1.1f%%",
       wedgeprops = {
           "edgecolor" : "black",
            "linewidth" : 1,
           "linestyle" : "--",
           "fill": False
       })
plt.title("expenses vs amounts")
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]
explode = [0,0.4,0,0.2,0]
plt.pie(amounts,labels = expenses,autopct = "%1.1f%%",explode = explode,startangle=180,shadow = True)
plt.title("expenses vs amounts")
```
