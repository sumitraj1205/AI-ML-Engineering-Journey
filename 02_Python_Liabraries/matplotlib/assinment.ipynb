```python
import matplotlib.pyplot as plt
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],  # New York
    [18, 19, 17, 20, 21],  # London
    [30, 32, 31, 33, 34],  # Delhi
    [25, 26, 24, 27, 28]   # Tokyo
]
a = int(input("enter no. of rows"))
b = int(input("enter no. of columns"))
fig,ax = plt.subplots(a,b)
k = 0
for i in range(a):
    j = 0
    for j in range(b):
        ax[i][j].plot(days,temperatures[k])
        ax[i][j].set_title(f"{cities[k]} plot")
        ax[i][j].set_xlabel("days")
        ax[i][j].set_ylabel("temperature")
        k += 1
    i += 1
        
fig.suptitle("days vs temperature")
fig.tight_layout()
```
