```python
import matplotlib.pyplot as plt
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]

bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]


plt.scatter(age,bp,label = "age vs bp",s = bp,marker = "<",alpha = 0.4,cmap="plasma",c = bp)
plt.xlabel("age")
plt.ylabel("bp")
plt.legend()
plt.colorbar(label = "bp")
for i in range(len(people)):
    plt.annotate(people[i],xy = (age[i],bp[i]),xytext = (age[i] +1,bp[i]))
plt.xlim(min(age),max(age) +5)
plt.ylim(min(bp),max(bp) +5)
```

```python
def solveNQueens(n):
    def backtrack(row, cols, diagonals, anti_diagonals, board):
        if row == n:
            result.append(["".join(r) for r in board])
            return
        
        for col in range(n):
            diag = row - col
            anti_diag = row + col
            
            if col in cols or diag in diagonals or anti_diag in anti_diagonals:
                continue
            
            # Place queen
            cols.add(col)
            diagonals.add(diag)
            anti_diagonals.add(anti_diag)
            board[row][col] = "Q"
            
            # Move to next row
            backtrack(row + 1, cols, diagonals, anti_diagonals, board)
            
            # Remove queen (backtrack)
            cols.remove(col)
            diagonals.remove(diag)
            anti_diagonals.remove(anti_diag)
            board[row][col] = "."
    
    result = []
    board = [["."] * n for _ in range(n)]
    backtrack(0, set(), set(), set(), board)
    return result
a = solveNQueens(3)
print(a)
```
