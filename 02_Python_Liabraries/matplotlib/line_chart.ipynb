```python
movies = ["The Dark Knight", "The Hurt Locker", "The King's Speech", "The Artist", "Argo"]

years = [2008, 2009, 2010, 2011, 2012]

oscar_revenue = [1005, 170, 427, 133, 232]
plt.plot(years,oscar_revenue)
plt.xlabel("years")
plt.ylabel("earnings in $")
```

```python
nonoscar_revenue = [265, 70, 127, 113, 32]
plt.plot(years,oscar_revenue,"<--r",label = "oscar movies")
plt.plot(years,nonoscar_revenue,"o-.b",label = "nonoscar movies")
plt.xlabel("years")
plt.ylabel("earnings in $")
plt.legend(loc = 'center')
```

```python
plt.plot(years,oscar_revenue,"<--b",label = "oscar movies")
plt.plot(years,nonoscar_revenue,linewidth = 5,color= "red",marker = "<",linestyle = "-.",label = "nonoscar movies")
plt.xlabel("years")
plt.ylabel("earnings in $")
plt.legend(loc = 'center')
plt.show()
```

```python
movies = ["The Dark Knight", "The Hurt Locker", "The King's Speech", "The Artist", "Argo"]

years = [2008, 2009, 2010, 2011, 2012]

oscar_revenue = [1005, 170, 427, 133, 232]
plt.bar(years,oscar_revenue)
plt.xlabel("years")
plt.ylabel("earnings in $")
plt.ylim(0,max(oscar_revenue) + 100)
for i in range(len(years)):
    plt.text(years[i],oscar_revenue[i] + 10,movies[i],ha = "center")
```
