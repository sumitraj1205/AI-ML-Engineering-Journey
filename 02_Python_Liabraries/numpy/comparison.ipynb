```python
import numpy as np
import time 
size = 10_000_000
a = list(range(size))
start = time .time()
sqrt = [x**2 for x in a]
end = time.time()
print(f"python list time = {end-start}")
b = np.array(a)
start = time .time()
sqrt = b ** 2
end = time.time()
print(f"numpy list time = {end-start}")
```
