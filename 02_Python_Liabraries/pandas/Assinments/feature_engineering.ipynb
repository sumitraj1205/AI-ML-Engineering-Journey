```python
#feature scalling
import pandas as pd
a = pd.read_csv("raw_data.csv")
b = a.copy()
```

```python
#apply functionc
b["tax"] = b["income"].apply(lambda x : "20%" if x>=60000 else "10%")
```

```python
#map function
c = {"Male":"M","Female":"F","Unknown":"U"}
b["gender"] = b["gender"].map(c)
```

```python
#assign function
b.assign(new_income = b["income"]*1.1)
```

```python
#replace function
b["country"] = b["country"].replace("USA","us")
```
