```python
#tranforming data
import pandas as pd
a = pd.read_csv("raw_data.csv")
b = a.copy()
```

```python
b.columns
```

```python
b.columns = ["Id","Name","age","Country", "Gender", "Income"]
```

```python
#rename function
b.rename(columns = {"Income":"Salary"})
b.rename(index = {2:"Second"})
```

```python
#sort values
b.sort_values("Income")
b["age"] = b["age"].fillna(b["age"].median())
b.sort_values(["Income","age"],ascending = False)
b.sort_index()
```

```python
#reset_index function
b.reset_index()
b.reset_index(drop = True)
```

```python
#rank function
b["Rankings"] = b["Income"].rank(ascending = False, method = "dense")
```

```python
#reorder
b[["Id","Name","age", "Gender", "Income","Country"]]
```
