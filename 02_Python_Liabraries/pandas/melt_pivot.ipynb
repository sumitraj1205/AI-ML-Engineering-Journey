```python
import pandas as pd
df = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22],
    "tax": [10, 20, 30, 40]
})
melted_data = df.melt(
    id_vars = ["country","year"],
    value_vars = ["sales","profit","tax"],
    var_name = "income",
    value_name = "value"
)

print(melted_data)
```

```python
original = melted_data.pivot(
    index = ["country","year"],
    columns = "income",
    values = "value"
)
print(original)
```
