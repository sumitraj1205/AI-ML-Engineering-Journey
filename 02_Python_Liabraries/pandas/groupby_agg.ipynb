```python
import pandas as pd
a = pd.read_csv("raw_data.csv")
b = a.copy()
```

```python
#groupby function
b.groupby("country")
b.groupby("country")["income"].mean()
b.groupby("gender")["income"].max()
```

```python
#agg function
b.groupby("gender")["income"].agg(["mean","min","max"])
```

```python
b.groupby("gender")["income"].agg(average_income = "mean",minimum_income = "min",maximum_income = "max")
```

```python
b.groupby("gender").agg({"income":"mean","age":"max"})
```

```python
b.groupby("gender").agg(maximum_salary = ("income","max"),average_age = ("age","mean"))
```
