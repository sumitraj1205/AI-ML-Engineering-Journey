```python
import pandas as pd
df1 = pd.DataFrame({
    "id" : [1,2,3],
    "name" : ["sumit","ravi","kunal"]
})
df2 = pd.DataFrame({
    "id" : [4,5,6],
    "name" : ["rohan","rahul","punit"]
})
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],ignore_index = True)
```
