```python
def is_sorted(arr, index):
    if index == len(arr) - 1:
        return True

    if arr[index] > arr[index + 1]:
        return False

    return is_sorted(arr, index + 1)


arr = [1, 2, 3, 4, 5]

if is_sorted(arr, 0):
    print("Array is sorted")
else:
    print("Array is not sorted")
```
